```python
%%writefile bank_analytics_main/metrics.py

def total_balance(accounts: list[dict]) -> float:
    return sum(acc["balance"] for acc in accounts)

def count_rich(accounts: list[dict], threshold: float) -> int:
    return sum(1 for acc in accounts if acc["balance"] > threshold)
```

```python
from bank_analytics_main.metrics import total_balance, count_rich

accounts = [
    {"client": "Aijan", "balance": 1200.0},
    {"client": "Nursultan", "balance": 500.0},
    {"client": "Sezim", "balance": 3000.0},
]

print("Total", total_balance(accounts))
print("Rich count (>1000):", count_rich(accounts, 2000))
```
